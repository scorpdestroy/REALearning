```python
small_set = np.array([10100,35500,105000,85000,25500,40500,65000])
small_set_median = sorted(small_set)[len(small_set)//2]
large_set = np.genfromtxt('income.csv', delimiter=',')
large_set_median = np.median(large_set)
print(small_set_median, large_set_median)
```

```python
patrons = np.array([2,6,14,4,3,9,1,11,4,2,8])
thirtieth_percentile = np.percentile(patrons, 30)
seventieth_percentile = np.percentile(patrons, 70)
print(thirtieth_percentile, seventieth_percentile)
```

```python
movies_watched = np.array([2,3,8,0,2,4,3,1,1,0,5,1,1,7,2])
first_quarter = np.percentile(movies_watched,25)
third_quarter = np.percentile(movies_watched,75)
interquartile_range = third_quarter-first_quarter
print(first_quarter, third_quarter, interquartile_range)
```

```python
pumpkin = np.array([68,1820,1420,2062,704,1156,1857,1755,2092,1384])
acorn_squash = np.array([20,43,99,200,12,250,58,120,230,215])
pumpkin_avg, acorn_squash_avg = np.mean(pumpkin), np.mean(acorn_squash)
pumpkin_std, acorn_squash_std = np.std(pumpkin), np.std(acorn_squash)
winner = 'pumpkin' if pumpkin_std > acorn_squash_std else 'acorn_squash'
print(pumpkin_avg, acorn_squash_avg, pumpkin_std, acorn_squash_std, winner)
```

```python
rainfall = np.array([5.21,3.76,3.27,2.35,1.89,1.55,.65,1.06,1.72,3.36,4.82,5.11])
rain_mean, rain_median = np.mean(rainfall), np.median(rainfall)
first_quarter, third_quarter = np.percentile(rainfall,[25,75])
interquartile_range = third_quarter-first_quarter
rain_std = np.std(rainfall)
print(rain_mean, rain_median, first_quarter, third_quarter, interquartile_range, rain_std)
```
