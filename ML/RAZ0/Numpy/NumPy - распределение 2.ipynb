```python
np.random.seed(42)
email = np.random.binomial(500, .05, 10000000)
plt.hist(email, bins=50)
plt.show
```

```python
no_emails = np.mean(email <= 4)

b_test_emails = np.mean(email >= 40)
print(no_emails)
print(b_test_emails)
```

```python
sunflowers = np.genfromtxt('sunflowers.csv', delimiter=',')
sunflowers_mean = np.mean(sunflowers)
sunflowers_std = np.std(sunflowers)
print(sunflowers_mean, sunflowers_std)

plt.hist(sunflowers,
range=(11, 15), histtype='step', linewidth=2,
label='observed')
plt.legend()
plt.show()
```

```python
sunflowers_normal = np.random.normal(sunflowers_mean, sunflowers_std, 5000)
plt.hist(sunflowers_normal,
range=(11, 15), histtype='step', linewidth=2,
label='normal')
plt.legend()
plt.show()
```
