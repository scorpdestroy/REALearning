```python
import numpy as np
from pathlib import Path
```

```python
np.random.seed(0)
flips = np.random.binomial(1, .5, 10000)
print(np.mean(flips))
```

```python
np.random.seed(0)
revenue = np.random.binomial(500, .05, 10000)
print(np.mean(revenue), np.std(revenue))
```

```python
sunflowers = np.genfromtxt('sunflowers.csv', delimiter=',') if Path('sunflowers.csv').exists() else np.array([13.89,14.7,13,11.37,12.1,11.82,13.09,13.78,12,12.83,12.1,11.09,13.32,11.86,14.35,11.7,12.85,14,13.83,12.02])
sunflowers_mean = np.mean(sunflowers)
sunflowers_std = np.std(sunflowers)
print(sunflowers_mean, sunflowers_std)
```
