```python
np.random.seed(0)
b_data = np.random.normal(6.7, .7, 1000)
f_data = np.random.normal(7.7, .3, 1000)
plt.hist(b_data, bins=30, range=(5,8.5), histtype='step')
plt.hist(f_data, bins=30, range=(5,8.5), histtype='step')
plt.show()
```

```python
np.random.seed(0)
email_opened = np.random.binomial(500, .05, 10000)
no_emails = np.mean(email_opened == 0)
b_test_emails = np.mean(email_opened >= 40)
print(no_emails, b_test_emails)
```
