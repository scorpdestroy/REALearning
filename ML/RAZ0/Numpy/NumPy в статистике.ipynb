```python
rainfall = np.array([5.21, 3.76, 3.27, 2.35, 1.89, 1.55, 0.65, 1.06,
                     1.72, 3.36, 4.82, 5.11])

rain_mean = np.mean(rainfall)
rain_median = np.median(rainfall)

first_quarter = np.percentile(rainfall, 25)
third_quarter = np.percentile(rainfall, 75)

interquartile_range = third_quarter - first_quarter

rain_std = np.std(rainfall)

print("Среднее:", rain_mean)
print("Медиана:", rain_median)
print("25-й процентиль (Q1):", first_quarter)
print("75-й процентиль (Q3):", third_quarter)
print("Межквартильный размах (IQR):", interquartile_range)
print("Стд. отклонение:", rain_std)
```
