```python
rng = np.random.default_rng(0)
a = rng.integers(1,11,10)
b = rng.integers(1,11,10)
print(a+b, a-b, a/b)
A = rng.integers(1,11,(3,3))
B = rng.integers(1,11,(3,3))
print(A @ B)
c = rng.integers(1,101,12).reshape(3,4)
print(c, c.shape)
print(c.reshape(4,3))
```

```python
rng = np.random.default_rng(0)
A = rng.integers(1,11,(4,4))
print(A, A.T)
C = rng.integers(1,11,(5,3))
print(np.linalg.matrix_rank(C), np.linalg.matrix_rank(C.T))
D = rng.integers(1,6,(4,5))
print(D, D.T, np.linalg.matrix_rank(D), np.linalg.matrix_rank(D.T))
```
