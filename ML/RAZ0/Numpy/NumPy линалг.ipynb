```python
rng = np.random.default_rng(0)
a = rng.integers(1,11,10)
b = rng.integers(1,11,10)
print(a+b, a-b, a/b)
A = rng.integers(1,11,(3,3))
B = rng.integers(1,11,(3,3))
print(A @ B)
c = rng.integers(1,101,12).reshape(3,4)
print(c, c.shape)
print(c.reshape(4,3))
```

```python
rng = np.random.default_rng(0)
A = rng.integers(1,11,(4,4))
print(A, A.T)
C = rng.integers(1,11,(5,3))
print(np.linalg.matrix_rank(C), np.linalg.matrix_rank(C.T))
D = rng.integers(1,6,(4,5))
print(D, D.T, np.linalg.matrix_rank(D), np.linalg.matrix_rank(D.T))
```

```python
from PIL import Image
import matplotlib.pyplot as plt

image = Image.open('cat_PNG50550.png')

image_array = np.array(image)
print("Форма массива:", image_array.shape)

gray_image_array = image_array[..., :3] @ np.array([0.299, 0.587, 0.114])

gray_image = Image.fromarray(gray_image_array.astype('uint8'), 'L')

new_size = (200, 200)

resized_image = gray_image.resize(new_size)

resized_image.save('processed_image.jpg')

plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.title('Исходное изображение')
plt.imshow(image)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title('Оттенки серого')
plt.imshow(gray_image, cmap='plasma')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('Измененный размер')
plt.imshow(resized_image, cmap='inferno')
plt.axis('off')

plt.tight_layout()
plt.show()
```
