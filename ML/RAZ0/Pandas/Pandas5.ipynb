```python
import pandas as pd

df = pd.DataFrame([
    ['John Smith', 15, 40],
    ['Jane Doe', 20, 43],
    ['Joe Schmo', 10, 35]
], columns=['name', 'hourly_wage', 'hours_worked'])
```

```python
mylambda = lambda x: x[0] + x[-1]
print(mylambda('This is a string'))
```

```python
mylambda = lambda age: "Добро пожаловать в BattleCity!" if age >= 13 else "Вам должно быть больше 13 лет"
mylambda(14)
```

```python
get_last_name = lambda x: x.split(' ')[-1]
df['last_name'] = df.name.apply(get_last_name)
df
```

```python
total_earned = lambda row: (row.hours_worked * row.hourly_wage) if row.hours_worked <= 40 else (40 * row.hourly_wage + (row.hours_worked - 40) * row.hourly_wage * 1.5)

df['total_earned'] = df.apply(total_earned, axis=1)
df
```
