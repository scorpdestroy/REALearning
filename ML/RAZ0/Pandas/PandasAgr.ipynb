```python
import pandas as pd
import numpy as np
orders = pd.read_csv('orders.csv', delimiter="\t")
```

```python
most_expensive = orders.price.max()
num_colors = orders.shoe_color.nunique()

orders.head(10)
```

```python
pricey_shoes = orders.groupby('shoe_type').price.max()
pricey_shoes
```

```python
type(pricey_shoes)
```

```python
pricey_shoes = orders.groupby('shoe_type').price.max().reset_index()
pricey_shoes
```

```python
type(pricey_shoes)
```

```python
cheap_shoes = orders.groupby('shoe_color').price.apply(lambda x: np.nanpercentile(x, 25)).reset_index()
cheap_shoes
```

```python
shoe_counts = orders.groupby(['shoe_type', 'shoe_color']).id.count().reset_index()
shoe_counts
```
