## Задание 1. OLS и GLS при автокорреляции

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy.linalg import toeplitz

np.random.seed(42)
n = 150
x1 = np.random.normal(size=n)
x2 = np.random.normal(size=n)
rho = 0.8
errors = np.zeros(n)
noise = np.random.normal(size=n)
for i in range(1, n):
    errors[i] = rho * errors[i - 1] + noise[i]
y = 5 + 2 * x1 - 3 * x2 + errors
X = sm.add_constant(np.column_stack([x1, x2]))
ols = sm.OLS(y, X).fit()
sigma = toeplitz(rho ** np.arange(n))
gls = sm.GLS(y, X, sigma=sigma).fit()
print(ols.summary())
print(gls.summary())
plt.figure(figsize=(10, 4))
plt.plot(ols.resid, label='OLS')
plt.plot(gls.resid, label='GLS')
plt.legend()
plt.show()
```

## Задание 2. OLS и GLS для Boston Housing

```python
boston = pd.read_csv('boston.csv')
print(boston.head())
plt.figure(figsize=(12, 9))
sns.heatmap(boston.corr().round(1), annot=True, cbar=False)
plt.show()
X_boston = sm.add_constant(boston.drop(columns='MEDV'))
y_boston = boston['MEDV']
boston_ols = sm.OLS(y_boston, X_boston).fit()
boston_gls = sm.GLS(y_boston, X_boston).fit()
print(boston_ols.summary())
print(boston_gls.summary())
```

## Задание 3. GLS для недвижимости

```python
house = pd.read_csv('Nedvig.csv', sep=';', encoding='cp1251').dropna()
X_house = pd.get_dummies(house.iloc[:, 1:-1], drop_first=True, dtype=float)
y_house = house.iloc[:, -1]
house_gls = sm.GLS(y_house, sm.add_constant(X_house)).fit()
print(house_gls.summary())
print('R2:', house_gls.rsquared)
```
