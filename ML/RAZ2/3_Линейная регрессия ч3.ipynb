## Задание 1. Lasso для data.csv

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Lasso, Ridge
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

data = pd.read_csv('data.csv')
X = data.drop(columns='y')
y = data['y']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
lasso = make_pipeline(StandardScaler(), Lasso(alpha=0.1, max_iter=10000))
lasso.fit(X_train, y_train)
predictions = lasso.predict(X_test)
print('Коэффициенты:', lasso[-1].coef_)
print('R2:', r2_score(y_test, predictions))
print('MSE:', mean_squared_error(y_test, predictions))
```

## Задание 2. Lasso для Boston Housing

```python
boston = pd.read_csv('boston.csv')
X = boston.drop(columns='MEDV')
y = boston['MEDV']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
boston_lasso = make_pipeline(StandardScaler(), Lasso(alpha=0.1, max_iter=10000))
boston_lasso.fit(X_train, y_train)
predictions = boston_lasso.predict(X_test)
print('Коэффициенты:', boston_lasso[-1].coef_)
print('R2:', r2_score(y_test, predictions))
print('MSE:', mean_squared_error(y_test, predictions))
```

## Задание 3. Ridge для Boston Housing

```python
boston_ridge = make_pipeline(StandardScaler(), Ridge(alpha=1.0))
boston_ridge.fit(X_train, y_train)
ridge_predictions = boston_ridge.predict(X_test)
print('Коэффициенты:', boston_ridge[-1].coef_)
print('R2:', r2_score(y_test, ridge_predictions))
print('MSE:', mean_squared_error(y_test, ridge_predictions))
```

## Итоговое задание. Lasso и Ridge для недвижимости

```python
house = pd.read_csv('Nedvig.csv', sep=';', encoding='cp1251').dropna()
house_test = pd.read_csv('Nedvig_test.csv', sep=';', encoding='cp1251')
X_house = pd.get_dummies(house.iloc[:, 1:-1], drop_first=True, dtype=float)
y_house = house.iloc[:, -1]
X_house_test = pd.get_dummies(house_test.iloc[:, 1:-1], drop_first=True, dtype=float)
X_house_test = X_house_test.reindex(columns=X_house.columns, fill_value=0)
y_house_test = house_test.iloc[:, -1]

models = {
    'Lasso': make_pipeline(StandardScaler(), Lasso(alpha=0.1, max_iter=10000)),
    'Ridge': make_pipeline(StandardScaler(), Ridge(alpha=1.0))
}
for name, model in models.items():
    model.fit(X_house, y_house)
    prediction = model.predict(X_house_test)
    print(name)
    print('Прогноз:', prediction)
    print('R2:', r2_score(y_house_test, prediction))
    print('MSE:', mean_squared_error(y_house_test, prediction))
```

## Ответы на контрольные вопросы

1. **Ridge-регрессия** — линейная регрессия с L2-регуляризацией, которая уменьшает большие коэффициенты, но обычно не обнуляет их.
2. **Lasso-регрессия** — линейная регрессия с L1-регуляризацией, способная обнулять коэффициенты и тем самым отбирать признаки.
3. Эти методы применяют при переобучении, большом числе признаков и мультиколлинеарности. Ridge полезна, когда большинство признаков значимы, а Lasso — когда нужен более компактный набор признаков.
