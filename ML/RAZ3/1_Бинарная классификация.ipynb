## Бинарная классификация диабета

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

data = pd.read_csv('diabetes.csv')
X = data.drop(columns='Outcome')
y = data['Outcome']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
prediction = model.predict(X_test)
probability = model.predict_proba(X_test)[:, 1]
print('Accuracy:', accuracy_score(y_test, prediction))
print('ROC-AUC:', roc_auc_score(y_test, probability))
print(classification_report(y_test, prediction))
ConfusionMatrixDisplay.from_predictions(y_test, prediction)
plt.show()
```

## Ответы на вопросы

1. Логистическая регрессия — алгоритм классификации, который оценивает вероятность принадлежности объекта к классу. Обычно её применяют для бинарной классификации.
2. Основная цель — получить вероятность класса и на её основе присвоить объекту метку.
3. Линейная регрессия прогнозирует непрерывное число, а логистическая — вероятность класса в диапазоне от 0 до 1.
4. Следует проверить независимость наблюдений, отсутствие сильной мультиколлинеарности, достаточный размер выборки, линейную связь числовых признаков с логарифмом шансов и влияние выбросов.
5. Коэффициент показывает изменение логарифма шансов при увеличении признака на единицу. Экспонента коэффициента даёт отношение шансов.
6. Сигмоида преобразует любое вещественное число в вероятность от 0 до 1: `1 / (1 + exp(-z))`.
7. Основные метрики: accuracy, precision, recall, F1, ROC-AUC, PR-AUC и матрица ошибок.
